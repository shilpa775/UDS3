### arbitary argument

####   syntax
-    def functionname(*varible_name):
-            variablename

```python
### arbitary argument

def add(*value):
    for i in value:
        print(i)
```

```python
add(1)
add(1,2)
add(1,2,3)
```

```python
def add_v3(a,b,*value):
    print(f"a value : {0}")
    print(f"a value : {b}")
    print(f"a value : {value}")
```

```python
add(1,2,3,1,2,4,5,6)
```

```python
add_22("hello",1,2,3,4)
```
