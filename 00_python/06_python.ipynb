#### while 
     - until the conditionn going to false it continune to run

```python
while True:
    print("hello")          ### it will print hello infinitely because the condition is always true. 

    ## this condition statics its always true because True is a truthy value. So it will print hello infinitely.
```

```python
start=0
while (start < 10):
    print("hello")          ### it will print hello infinitely because the condition is always true.
```

```python
start=0
while (start < 10):
    print("hello" , start) 
    start+=1         
```

```python
start=0
print("while loop is started : start")
while (start < 10):
    print("hello" , start) 
    start+=1                           ## condition to make it false afteer 10 itreation 
print("while loop is ended : end ")
```

```python
start = 0
stop=True
while stop:
    val=float(input("enter value you want to add or type stop to exit"))
    start+=val
    print(start)
    user_continue=input("do you wish to continue y for yes , n for not") .lower()
    if user_continue=='n':
       stop=False
```
