### slicing
    - extracting part of sequence

    - sequence(start,stop,step)

    - start-> index to begin from,include
    - stop-> index to end , exclusive
    - step-> how many position to move each time
    - sequence-> varisble

```python
# slicing
student=["student_1","student_2","student_3","student_4","student_5","student_6","student_7","student_8"]
print(student[0:3])
print(student[-1:3]) # going left to right
print(student[-3:-1]) # going to left to right
```

```python
print(student[-1:3:-1])
print(student[-1:-3:1])
print(student[2:-3])
print(student[0:-5:-4])
```

```python
# reversed excluiding 0 index value
a=[1,2,3,4,5,6,7]
print(a[-1:0:-1])
```

```python
## its also revesed including all the value
a[::-1]
```

```python
print(a[-1::-1])
print(a[::-1])
```

```python
b=[0,1,2,3,4,[0,1,2,3]] # list inside list
print(b[0])
print(b[3])
print(b[5])
print(b[5][2])
```
